# pylint Tool Runner

This notebook runs the Python-only Lint / Rule Violation setup using `run_pylint.py`.


```python
from pathlib import Path
import subprocess
import sys

ROOT = Path.cwd()
RUNNER = ROOT / "run_pylint.py"
DEFAULT_REPO = ROOT.parent / "Code Duplication" / "github-actions-cicd-example"
DEFAULT_OUTPUT = ROOT / "pylint_report"

def run_command(args):
    result = subprocess.run(args, cwd=ROOT, capture_output=True, text=True)
    if result.stdout:
        print(result.stdout)
    if result.stderr:
        print(result.stderr, file=sys.stderr)
    if result.returncode != 0:
        raise RuntimeError(f"Command failed with exit code {result.returncode}")
    return result.returncode

print(f"Runner: {RUNNER}")
print(f"Default repo: {DEFAULT_REPO}")
print(f"Default output: {DEFAULT_OUTPUT}")
```

## Install pylint


```python
run_command([
    sys.executable,
    str(RUNNER),
    "install",
])
```

## Configure Target Repo


```python
repo_path = DEFAULT_REPO
output_path = DEFAULT_OUTPUT
repo_path, output_path
```

## Run pylint Text Output


```python
run_command([
    sys.executable,
    str(RUNNER),
    "run",
    "--repo",
    str(repo_path),
    "--output",
    str(output_path / "pylint_text.txt"),
    "--",
    "--exit-zero",
    "--score",
    "y",
    "actionscicd",
    "tests",
])
```

## Run pylint JSON Output


```python
run_command([
    sys.executable,
    str(RUNNER),
    "run",
    "--repo",
    str(repo_path),
    "--output",
    str(output_path / "pylint_json2.json"),
    "--",
    "--exit-zero",
    "--output-format=json2",
    "actionscicd",
    "tests",
])
```

## Inspect Saved Outputs


```python
print((output_path / "pylint_text.txt").read_text(encoding="utf-8"))
print((output_path / "pylint_json2.json").read_text(encoding="utf-8")[:1500])
```
