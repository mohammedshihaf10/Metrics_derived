# Coverage.py Tool Runner

This notebook is separate from the current setup and uses the existing `run_tool.py` script.


```python
from pathlib import Path
import subprocess
import sys

ROOT = Path.cwd()
RUNNER = ROOT / "run_tool.py"
DEFAULT_REPO = ROOT.parent / "github-actions-cicd-example"

def run_command(args):
    result = subprocess.run(
        args,
        cwd=ROOT,
        capture_output=True,
        text=True,
    )
    if result.stdout:
        print(result.stdout)
    if result.stderr:
        print(result.stderr, file=sys.stderr)
    if result.returncode != 0:
        raise RuntimeError(f"Command failed with exit code {result.returncode}")
    return result.returncode

print(f"Runner: {RUNNER}")
print(f"Default repo: {DEFAULT_REPO}")
```

## Install Coverage.py

Runs the existing isolated coverage.py install flow in `tools/coveragepy`.


```python
run_command([
    sys.executable,
    str(RUNNER),
    "install",
])
```

## Configure Target Repo

Change `repo_path` only when you want to point coverage.py at a different repository.


```python
repo_path = DEFAULT_REPO
repo_path
```

## Show Raw Data

Displays the raw TSV file used by the example package.


```python
raw_data_path = repo_path / "actionscicd" / "data" / "mogi_data.tsv"
raw_data = raw_data_path.read_text(encoding="utf-8")
print(raw_data)
```

## Run Coverage.py Commands


```python
run_command([
    sys.executable,
    str(RUNNER),
    "run",
    "--repo",
    str(repo_path),
    "--",
    "run",
    "-m",
    "unittest",
    "discover",
])
```

```python
run_command([
    sys.executable,
    str(RUNNER),
    "run",
    "--repo",
    str(repo_path),
    "--",
    "report",
])
```

```python
run_command([
    sys.executable,
    str(RUNNER),
    "run",
    "--repo",
    str(repo_path),
    "--",
    "xml",
])
```
