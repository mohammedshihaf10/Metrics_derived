# pymcdc Tool Runner

This notebook uses the existing `run_pymcdc.py` script.


```python
from pathlib import Path
import subprocess
import sys

ROOT = Path.cwd()
RUNNER = ROOT / "run_pymcdc.py"
TARGET = ROOT / "tool_examples" / "logic_examples.py"

def run_command(args):
    result = subprocess.run(
        args,
        cwd=ROOT,
        capture_output=True,
        text=True,
    )
    if result.stdout:
        print(result.stdout)
    if result.stderr:
        print(result.stderr, file=sys.stderr)
    if result.returncode != 0:
        raise RuntimeError(f"Command failed with exit code {result.returncode}")
    return result.returncode

print(f"Runner: {RUNNER}")
print(f"Target file: {TARGET}")
```

## Install pymcdc


```python
run_command([
    sys.executable,
    str(RUNNER),
    "install",
])
```

## Run pymcdc


```python
run_command([
    sys.executable,
    str(RUNNER),
    "run",
    "--repo",
    str(ROOT),
    "--",
    str(TARGET),
])
```
