# testmon Tool Runner

This notebook uses the existing `run_testmon.py` script.


```python
from pathlib import Path
import subprocess
import sys

ROOT = Path.cwd()
RUNNER = ROOT / "run_testmon.py"
REPO = ROOT.parent / "github-actions-cicd-example"
TESTMON_DATA = REPO / ".testmondata"

def run_command(args):
    result = subprocess.run(
        args,
        cwd=ROOT,
        capture_output=True,
        text=True,
    )
    if result.stdout:
        print(result.stdout)
    if result.stderr:
        print(result.stderr, file=sys.stderr)
    if result.returncode != 0:
        raise RuntimeError(f"Command failed with exit code {result.returncode}")
    return result.returncode

print(f"Runner: {RUNNER}")
print(f"Repo: {REPO}")
```

## Install testmon


```python
run_command([
    sys.executable,
    str(RUNNER),
    "install",
])
```

## Run testmon twice

The second run demonstrates that unchanged code can result in zero selected tests.


```python
if TESTMON_DATA.exists():
    TESTMON_DATA.unlink()

run_command([
    sys.executable,
    str(RUNNER),
    "run",
    "--repo",
    str(REPO),
    "--",
    "-q",
    "tests/test_somecode.py",
])

run_command([
    sys.executable,
    str(RUNNER),
    "run",
    "--repo",
    str(REPO),
    "--",
    "-q",
    "tests/test_somecode.py",
])
```
