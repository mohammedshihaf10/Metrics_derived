# flake8-cognitive-complexity Tool Runner


```python
from pathlib import Path
import subprocess
import sys

ROOT = Path.cwd()
RUNNER = ROOT / "run_flake8cog.py"
REPO = ROOT.parent / "github-actions-cicd-example"
TARGET = REPO / "actionscicd" / "somecode.py"

def run_command(args):
    result = subprocess.run(args, cwd=ROOT, capture_output=True, text=True)
    if result.stdout:
        print(result.stdout)
    if result.stderr:
        print(result.stderr, file=sys.stderr)
    if result.returncode != 0 and result.returncode != 1:
        raise RuntimeError(f"Command failed with exit code {result.returncode}")
    return result.returncode
```

```python
run_command([sys.executable, str(RUNNER), "install"])
```

```python
run_command([sys.executable, str(RUNNER), "run", "--repo", str(REPO), "--", "--extend-ignore=E501", "--max-cognitive-complexity=1", str(TARGET)])
```
