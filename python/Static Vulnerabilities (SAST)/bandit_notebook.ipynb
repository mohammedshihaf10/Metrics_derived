# Bandit Tool Runner

This notebook runs the Python-only Static Vulnerabilities (SAST) setup using `run_bandit.py`.


```python
from pathlib import Path
import subprocess
import sys

ROOT = Path.cwd()
RUNNER = ROOT / "run_bandit.py"
DEFAULT_REPO = ROOT.parent / "Code Duplication" / "github-actions-cicd-example"
DEFAULT_OUTPUT = ROOT / "bandit_report" / "bandit.json"

def run_command(args):
    result = subprocess.run(args, cwd=ROOT, capture_output=True, text=True)
    if result.stdout:
        print(result.stdout)
    if result.stderr:
        print(result.stderr, file=sys.stderr)
    if result.returncode not in (0, 1):
        raise RuntimeError(f"Command failed with exit code {result.returncode}")
    return result.returncode

print(f"Runner: {RUNNER}")
print(f"Default repo: {DEFAULT_REPO}")
print(f"Default output: {DEFAULT_OUTPUT}")
```

## Install Bandit


```python
run_command([
    sys.executable,
    str(RUNNER),
    "install",
])
```

## Configure Target Repo


```python
repo_path = DEFAULT_REPO
output_path = DEFAULT_OUTPUT
repo_path, output_path
```

## Run Bandit JSON Output


```python
run_command([
    sys.executable,
    str(RUNNER),
    "run",
    "--repo",
    str(repo_path),
    "--output",
    str(output_path),
    "--",
    "-r",
    "hashes",
    "ciphers",
    "other",
    "web_programming",
    "-f",
    "json",
    "-lll",
    "-iii",
])
```

## Inspect Saved Output


```python
import json

report = json.loads(output_path.read_text(encoding="utf-8"))
print(report["metrics"]["_totals"])
print(report["results"][:5])
```
