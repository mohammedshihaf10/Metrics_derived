# coverage.py Branch Coverage


```python
from pathlib import Path
import subprocess, sys
ROOT = Path.cwd()
RUNNER = ROOT / 'run_coveragepy.py'
REPO = ROOT.parent / 'Cyclomatic Complexity' / 'github-actions-cicd-example'
```

```python
subprocess.run([sys.executable, str(RUNNER), 'install'], cwd=ROOT, check=True)
```

```python
subprocess.run([sys.executable, str(RUNNER), 'run', '--repo', str(REPO), '--', 'erase'], cwd=ROOT, check=True)
subprocess.run([sys.executable, str(RUNNER), 'run', '--repo', str(REPO), '--', 'run', '--branch', '-m', 'pytest'], cwd=ROOT, check=True)
subprocess.run([sys.executable, str(RUNNER), 'run', '--repo', str(REPO), '--', 'json', '-o', 'coverage_branch.json'], cwd=ROOT, check=True)
subprocess.run([sys.executable, str(RUNNER), 'run', '--repo', str(REPO), '--', 'xml', '-o', 'coverage_branch.xml'], cwd=ROOT, check=True)
subprocess.run([sys.executable, str(RUNNER), 'run', '--repo', str(REPO), '--', 'report'], cwd=ROOT, check=True)
```
