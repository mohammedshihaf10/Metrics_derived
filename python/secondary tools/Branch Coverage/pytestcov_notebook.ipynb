# pytest-cov Branch Coverage


```python
from pathlib import Path
import subprocess, sys, json
ROOT = Path.cwd()
RUNNER = ROOT / 'run_pytestcov.py'
REPO = (ROOT.parent / "github-actions-cicd-example").resolve()
```

```python
subprocess.run([sys.executable, str(RUNNER), 'install'], cwd=ROOT, check=True)
```

```python
subprocess.run([sys.executable, str(RUNNER), 'run', '--repo', str(REPO), '--', '--cov=actionscicd', '--cov-branch', '--cov-report=json:coverage_branch_secondary.json', '-q'], cwd=ROOT, check=True)
```

```python
print(json.loads((REPO / 'coverage_branch_secondary.json').read_text(encoding='utf-8'))['totals'])
```
