# PyDriller Tool Runner

This notebook runs the Python-only Code Churn setup using `run_pydriller.py`.


```python
from pathlib import Path
import subprocess
import sys

ROOT = Path.cwd()
RUNNER = ROOT / "run_pydriller.py"
DEFAULT_REPO = ROOT.parent / "Cyclomatic Complexity" / "github-actions-cicd-example"
DEFAULT_OUTPUT = ROOT / "pydriller_report" / "code_churn.json"

def run_command(args):
    result = subprocess.run(args, cwd=ROOT, capture_output=True, text=True)
    if result.stdout:
        print(result.stdout)
    if result.stderr:
        print(result.stderr, file=sys.stderr)
    if result.returncode != 0:
        raise RuntimeError(f"Command failed with exit code {result.returncode}")
    return result.returncode

print(f"Runner: {RUNNER}")
print(f"Default repo: {DEFAULT_REPO}")
print(f"Default output: {DEFAULT_OUTPUT}")
```

## Install PyDriller


```python
run_command([
    sys.executable,
    str(RUNNER),
    "install",
])
```

## Configure Target Repo


```python
repo_path = DEFAULT_REPO
output_path = DEFAULT_OUTPUT
repo_path, output_path
```

## Run Code Churn Analysis


```python
run_command([
    sys.executable,
    str(RUNNER),
    "run",
    "--repo",
    str(repo_path),
    "--output",
    str(output_path),
])
```

## Inspect Saved Output


```python
import json

report = json.loads(output_path.read_text(encoding="utf-8"))
print(report['summary'])
print(report['top_files'][:5])
```
