# jscpd Tool Runner

This notebook runs the Python-only Code Duplication setup using `run_jscpd.py`.

```python
from pathlib import Path
import subprocess
import sys

ROOT = Path.cwd()
RUNNER = ROOT / "run_jscpd.py"
DEFAULT_REPO = ROOT / "github-actions-cicd-example"
DEFAULT_OUTPUT = ROOT / "jscpd_report"

def run_command(args, cwd=None):
    result = subprocess.run(args, cwd=cwd or ROOT, capture_output=True, text=True)
    if result.stdout:
        print(result.stdout)
    if result.stderr:
        print(result.stderr, file=sys.stderr)
    if result.returncode != 0:
        raise RuntimeError(f"Command failed with exit code {result.returncode}")
    return result.returncode

print(f"Runner: {RUNNER}")
print(f"Default repo: {DEFAULT_REPO}")
print(f"Default output: {DEFAULT_OUTPUT}")
```

## Install jscpd

Runs the isolated local npm install flow in `tools/jscpd`.

```python
run_command([
    sys.executable,
    str(RUNNER),
    "install",
])
```

## Configure Target Repo


```python
repo_path = DEFAULT_REPO
output_path = DEFAULT_OUTPUT
repo_path, output_path
```

## Run jscpd


```python
run_command([
    sys.executable,
    str(RUNNER),
    "run",
    "--repo",
    str(repo_path),
    "--output",
    str(output_path),
    "--",
    "--format",
    "python",
    "--pattern",
    "**/*.py",
    "--min-lines",
    "5",
    "--min-tokens",
    "30",
    "--reporters",
    "console,json",
    ".",
], cwd=repo_path)
```

## Inspect JSON Report


```python
import json

report = json.loads((output_path / 'jscpd-report.json').read_text(encoding='utf-8'))
print(report['statistics']['total'])
print(report['duplicates'][:3])
```
