# Radon Tool Runner

This notebook is separate from the current setup and uses the existing `run_radon.py` script.


```python
from pathlib import Path
import subprocess
import sys

ROOT = Path.cwd()
RUNNER = ROOT / "run_radon.py"
DEFAULT_REPO = ROOT / "github-actions-cicd-example"

def run_command(args):
    result = subprocess.run(
        args,
        cwd=ROOT,
        capture_output=True,
        text=True,
    )
    if result.stdout:
        print(result.stdout)
    if result.stderr:
        print(result.stderr, file=sys.stderr)
    if result.returncode != 0:
        raise RuntimeError(f"Command failed with exit code {result.returncode}")
    return result.returncode

print(f"Runner: {RUNNER}")
print(f"Default repo: {DEFAULT_REPO}")
```

## Install Radon

Runs the existing isolated radon install flow in `tools/radon`.


```python
run_command([
    sys.executable,
    str(RUNNER),
    "install",
])
```

## Configure Target Repo

Change `repo_path` only when you want to point radon at a different repository.


```python
repo_path = DEFAULT_REPO
repo_path
```

## Run Radon Commands


```python
run_command([
    sys.executable,
    str(RUNNER),
    "run",
    "--repo",
    str(repo_path),
    "--",
    "cc",
    ".",
    "-s",
])
```
