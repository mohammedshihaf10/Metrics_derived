# complexipy Tool Runner

This notebook uses the existing `run_complexipy.py` script.


```python
from pathlib import Path
import subprocess
import sys

ROOT = Path.cwd()
RUNNER = ROOT / "run_complexipy.py"
REPO = ROOT / "github-actions-cicd-example"
TARGET = REPO / "actionscicd" / "somecode.py"

def run_command(args):
    result = subprocess.run(args, cwd=ROOT, capture_output=True, text=True)
    if result.stdout:
        print(result.stdout)
    if result.stderr:
        print(result.stderr, file=sys.stderr)
    if result.returncode != 0:
        raise RuntimeError(f"Command failed with exit code {result.returncode}")
    return result.returncode
```

```python
run_command([sys.executable, str(RUNNER), "install"])
```

```python
run_command([
    sys.executable,
    str(RUNNER),
    "run",
    "--repo",
    str(REPO),
    "--",
    str(TARGET),
    "--ignore-complexity",
    "--sort",
    "desc",
    "--color",
    "no",
])
```

```python
run_command([
    sys.executable,
    str(RUNNER),
    "run",
    "--repo",
    str(REPO),
    "--",
    str(TARGET),
    "--ignore-complexity",
    "--sort",
    "desc",
    "--output-json",
    "--color",
    "no",
])
```
